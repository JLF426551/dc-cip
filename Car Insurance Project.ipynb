### Project Instructions

*Tasks*
* Identify the single feature of the data that is the best predictor of whether a customer will put in a claim (`outcome` = 1 )
* Store as a DataFrame called `best_feature_df`, containing the columns named `best_feature` and `best_accuracy` with the name of the feature with the highest accuracy and the respective accuracy score.

**Set-Up**

Load files and import libraries

```python
# Import required modules
import pandas
import numpy
from statsmodels.formula.api import logit
```

```python
mData = pandas.read_csv('car_insurance.csv')
```

```python
# For features to be texted extract all columns from dataframe
# First and last items should not be included (correspond to 'id' and 'outcome')
mColumns = mData.columns[1:-1]
```

**Steps / Explanation**

* Iterate through columns
* Build a model for each
* Keep track of variables fed
* Save information on best performing model

```python
#Fit Model
testModel = logit(' outcome ~ age', data = mData).fit()

#Set-Up Matrix
# Actual Values
actual_values = mData['outcome']

# predicted values
#predicted_values = numpy.round( testModel.predict())

#Save as single dataframe
#outcomes = pandas.DataFrame( {"actual": actual_values, 
#                              "predicted": predicted_values })

conf_matrix = testModel.pred_table()

# Extract TN, TP, FN and FP from conf_matrix
TN = conf_matrix[0,0]
TP = conf_matrix[1,1]
FN = conf_matrix[1,0]
FP = conf_matrix[0,1]

# Calculate and print the accuracy
accuracy = (TN + TP) / (TN + FN + FP + TP)
```

```python
print(accuracy)
```

```python
sColumns = mColumns[:3]
```

```python
mBestFeatureName = ""
mBestAccuracy = 0.00

for feature_name in mColumns:
    promptString = "outcome ~ " + feature_name
    testModel = logit(promptString , data = mData).fit()
    conf_matrix = testModel.pred_table()

    TN = conf_matrix[0,0]
    TP = conf_matrix[1,1]
    FN = conf_matrix[1,0]
    FP = conf_matrix[0,1]

    tmpAccuracy = (TN + TP) / (TN + FN + FP + TP)
    #Optional output, used as part of debugging. Keeping to visualize resuls as they're processed.
    print("Accuracy for " + feature_name + " is:" + str(tmpAccuracy) )

    if tmpAccuracy > mBestAccuracy:
        mBestAccuracy = tmpAccuracy
        mBestFeatureName = feature_name
```

**Best Feature and Accuracy**

```python
#Output Results to Screen
print("Best predicting feature is " + mBestFeatureName + " with an accuracy of:" + str(round(mBestAccuracy,4)) )
```

```python
#Store as dataframe

best_feature_df = pandas.DataFrame(data = {"best_feature":mBestFeatureName, 
                               "best_accuracy":mBestAccuracy}, index = [0])
```
